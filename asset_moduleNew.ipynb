```python
%%writefile asset_module
from dataclasses import dataclass


@dataclass
class SubseaJumper:
    name: str
    design_pressure_bar: float
    operating_pressure_bar: float
    tension_capacity_kn: float
    applied_tension_kn: float

    def utilisation(self) -> float:
        """Return the governing pressure/tension utilisation ratio."""
        pressure_ratio = self.operating_pressure_bar / self.design_pressure_bar
        tension_ratio = self.applied_tension_kn / self.tension_capacity_kn
        return max(pressure_ratio, tension_ratio)

    def status_report(self) -> str:
        utilisation_percent = self.utilisation() * 100
        status = "PASS" if self.utilisation() <= 1.0 else "OVER LIMIT"
        return f"{self.name}: {status} - utilisation {utilisation_percent:.1f}%"


@dataclass
class RiserSystem:
    name: str
    tension_capacity_kn: float
    applied_tension_kn: float
    bending_capacity_knm: float
    applied_bending_knm: float

    def utilisation(self) -> float:
        """Return the governing tension/bending utilisation ratio."""
        tension_ratio = self.applied_tension_kn / self.tension_capacity_kn
        bending_ratio = self.applied_bending_knm / self.bending_capacity_knm
        return max(tension_ratio, bending_ratio)

    def status_report(self) -> str:
        utilisation_percent = self.utilisation() * 100
        status = "PASS" if self.utilisation() <= 1.0 else "OVER LIMIT"
        return f"{self.name}: {status} - utilisation {utilisation_percent:.1f}%"


jumper = SubseaJumper(
    "Jumper-J01", 350, 280, 1200, 800
)

riser = RiserSystem(
    "Riser-R01", 2000, 1700, 900, 500
)

print(jumper.status_report())
print(riser.status_report())
```
