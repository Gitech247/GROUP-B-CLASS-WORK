```python
#Question 2.
def fahrenheit_to_celsius(fahrenheit: float) -> float:
    """
    Convert temperature from Fahrenheit to Celsius.

    Args:
        fahrenheit: Temperature in degrees Fahrenheit.

    Returns:
        Temperature in degrees Celsius.

    Raises:
        TypeError: If the input is not a number.
    """
    if not isinstance(fahrenheit, (int, float)) or isinstance(fahrenheit, bool):
        raise TypeError("Temperature must be a number.")

    return (fahrenheit - 32) * 5 / 9


# Test 1
print("30°F =", fahrenheit_to_celsius(32), "°C")

# Test 2
print("212°F =", fahrenheit_to_celsius(212), "°C")

# Test 3: Invalid input
try:
    print(fahrenheit_to_celsius("100"))
except TypeError as e:
    print("Error:", e)
```

```python
#**Question 6**
import math

class SubseaPipeline:
    def __init__(self, length_km, od_inches, wt_mm, material_grade):
        self.length_km = length_km
        self.od_inches = od_inches
        self.wt_mm = wt_mm
        self.material_grade = material_grade

    def pipe_volume_m3(self):
        """
        Calculate the internal volume of the pipeline in cubic metres.
        """
        # Convert units
        length_m = self.length_km * 1000
        od_m = self.od_inches * 0.0254
        wt_m = self.wt_mm / 1000

        # Calculate internal diameter
        id_m = od_m - (2 * wt_m)

        # Calculate internal volume
        volume = math.pi * (id_m / 2) ** 2 * length_m

        return volume


# Create a subsea pipeline object
pipeline = SubseaPipeline(
    length_km=10,
    od_inches=10.75,
    wt_mm=9.38,
    material_grade="X65"
)

# Calculate volume
volume = pipeline.pipe_volume_m3()

print(f"Pipeline internal volume: {volume:.2f} m³")
```

Modern Excel worksheets have a maximum of 1,048,576 rows. Our annual SCADA dataset contains:

$$ 157,680,000 \text{ data points} $$

**Question 9**
The try/except pattern is used for handling errors without crashing the program.

>try → contains code that might cause an error.
except → runs when a specific error occurs.
>This is useful when working with engineering data where unexpected values can occur.

```python
#code
def safe_divide(numerator, denominator):
    """
    Safely divide two numbers.

    Returns:
        The division result, or an error message if
        the denominator is zero.
    """
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return "Error: Division by zero"


# Test the function
print(safe_divide(100, 5))
print(safe_divide(100, 0))
```
